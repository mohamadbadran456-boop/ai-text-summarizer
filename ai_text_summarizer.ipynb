AI Text Summarizer

Objective:
Build a simple AI application that summarizes English text
using a pretrained model.

Tools:
Python, Google Colab, and Hugging Face Transformers.

Expected result:
The user enters a paragraph and receives a shorter summary.
The summary is checked against the original text for accuracy.

```python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

model_name = "google-t5/t5-small"

tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSeq2SeqLM.from_pretrained(model_name)
model.eval()

print("Model is ready!")
```

```python
text = """
A city launched a recycling program last month.
Residents received separate bins for paper, plastic, and glass.
The city organized workshops to explain how to sort waste.
More than 500 families joined the program in its first week.
Officials hope to reduce waste sent to landfills.
"""

inputs = tokenizer(
    "summarize: " + text,
    return_tensors="pt",
    max_length=512,
    truncation=True
)

outputs = model.generate(
    **inputs,
    max_new_tokens=70,
    num_beams=4,
    do_sample=False
)

summary = tokenizer.decode(outputs[0], skip_special_tokens=True)

print("Original text:\n", text)
print("\nSummary:\n", summary)
```

Testing Results

Test 1 — School Library:
The summary included correct details about the facilities,
but missed the opening of the new library.

Test 2 — Recycling Program:
The summary mentioned separate bins and participation by
more than 500 families, but missed the program's launch and goal.

Conclusion:
The application generates shorter summaries, but it can miss
important ideas. Human review is necessary.
